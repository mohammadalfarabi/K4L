# Perkalian Silang (Cross Product)

Perkalian silang atau cross product adalah operasi matematika antara dua vektor tiga dimensi yang menghasilkan vektor baru yang tegak lurus (orthogonal) terhadap kedua vektor asal. Operasi ini hanya berlaku untuk vektor di ruang tiga dimensi (3D).

### Diberikan:

$$
\vec{A} = \begin{bmatrix} A_x \\ A_y \\ A_z \end{bmatrix}, \quad \vec{B} = \begin{bmatrix} B_x \\ B_y \\ B_z \end{bmatrix}
$$

### Rumus:

$$
\vec{A} \times \vec{B} =
\begin{bmatrix}
A_yB_z - A_zB_y \\
-(A_xB_z - A_zB_x) \\
A_xB_y - A_yB_x
\end{bmatrix}
$$

Contoh:

$$
\vec{A} = \begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix}, \quad \vec{B} = \begin{bmatrix} 4 \\ 5 \\ 6 \end{bmatrix}
$$

$$
\vec{A} \times \vec{B} =
\begin{bmatrix}
(2)(6) - (3)(5) \\
-((1)(6) - (3)(4)) \\
(1)(5) - (2)(4)
\end{bmatrix}
=
\begin{bmatrix}
-3 \\
6 \\
-3
\end{bmatrix}
$$


## Menggunakan determinan untuk menghitung perkalian silang (cross product)

Misalnya:

$$
\vec{a} = \begin{bmatrix} a_1 \\ a_2 \\ a_3 \end{bmatrix}, \quad
\vec{b} = \begin{bmatrix} b_1 \\ b_2 \\ b_3 \end{bmatrix}
$$

Maka, perkalian silang $\vec{a} \times \vec{b}$ dapat dihitung menggunakan **determinan matriks 3x3**:

$$
\vec{a} \times \vec{b} =
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
a_1 & a_2 & a_3 \\
b_1 & b_2 & b_3
\end{vmatrix}
$$

### Langkah-langkah:

Hitung determinan dengan ekspansi baris pertama:

$$
\vec{a} \times \vec{b} =
\hat{i}(a_2 b_3 - a_3 b_2) - \hat{j}(a_1 b_3 - a_3 b_1) + \hat{k}(a_1 b_2 - a_2 b_1)
$$

Atau dalam bentuk vektor:

$$
\vec{a} \times \vec{b} =
\begin{bmatrix}
a_2 b_3 - a_3 b_2 \\
-(a_1 b_3 - a_3 b_1) \\
a_1 b_2 - a_2 b_1
\end{bmatrix}
$$

### Contoh:

Misal:

$$
\vec{a} = \begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix}, \quad
\vec{b} = \begin{bmatrix} 4 \\ 5 \\ 6 \end{bmatrix}
$$

Maka:

$$
\vec{a} \times \vec{b} =
\begin{bmatrix}
(2)(6) - (3)(5) \\
-(1)(6) - (3)(4) \\
(1)(5) - (2)(4)
\end{bmatrix}
=
\begin{bmatrix}
12 - 15 \\
-(6 - 12) \\
5 - 8
\end{bmatrix}
=
\begin{bmatrix}
-3 \\
6 \\
-3
\end{bmatrix}
$$

## Hasil Kali Silang dan Sudut

Jika $\vec{u}$ dan $\vec{v}$ adalah vektor tak nol di $\mathbb{R}^3$, maka:

$$
\|\vec{u} \times \vec{v}\| = \|\vec{u}\| \cdot \|\vec{v}\| \cdot \sin(\theta)
$$

di mana:

* $\|\vec{u} \times \vec{v}\|$ adalah magnitudo (panjang) dari hasil cross product
* $\|\vec{u}\|$ dan $\|\vec{v}\|$ adalah panjang masing-masing vektor
* $\theta$ adalah sudut antara vektor $\vec{u}$ dan $\vec{v}$, dengan:

  $$
  0 \leq \theta \leq \pi \quad (0^\circ \text{ sampai } 180^\circ)
  $$

Makna Geometris:

* Cross product menghasilkan vektor yang:

  * Tegak lurus terhadap $\vec{u}$ dan $\vec{v}$
  * Panjangnya sama dengan luas jajargenjang yang dibentuk oleh $\vec{u}$ dan $\vec{v}$


## Implementasi luas jajar genjang dengan menggunakan perkalian silang vektor.


### Rumus Umum:

Misalkan dua vektor $\vec{a}$ dan $\vec{b}$ membentuk dua sisi jajargenjang yang saling menempel dari satu titik sudut. Maka:

$$
\text{Luas Jajargenjang} = \|\vec{a} \times \vec{b}\|
$$

### Langkah-Langkah Perhitungan:

1. Hitung $\vec{a} \times \vec{b}$ (perkalian silang biasa)
2. Hitung panjang (magnitudo) dari hasil tersebut:

   $$
   \|\vec{a} \times \vec{b}\| = \sqrt{x^2 + y^2 + z^2}
   $$

### Contoh Perhitungan:

Misal:

$$
\vec{a} = \begin{bmatrix} 2 \\ 3 \\ 1 \end{bmatrix}, \quad
\vec{b} = \begin{bmatrix} 1 \\ 0 \\ 4 \end{bmatrix}
$$

Langkah 1: Hitung cross product $\vec{a} \times \vec{b}$

$$
\vec{a} \times \vec{b} =
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
2 & 3 & 1 \\
1 & 0 & 4
\end{vmatrix}
=
\hat{i}(3 \cdot 4 - 1 \cdot 0)
- \hat{j}(2 \cdot 4 - 1 \cdot 1)
+ \hat{k}(2 \cdot 0 - 3 \cdot 1)
$$

$$
= \hat{i}(12) - \hat{j}(8 - 1) + \hat{k}(-3)
= \begin{bmatrix} 12 \\ -7 \\ -3 \end{bmatrix}
$$

Langkah 2: Hitung panjang vektor hasilnya

$$
\|\vec{a} \times \vec{b}\| =
\sqrt{12^2 + (-7)^2 + (-3)^2} = \sqrt{144 + 49 + 9} = \sqrt{202}
$$

Jadi, luas jajargenjang = $\sqrt{202}$ satuan luas.

## Volume paralelepiped (balok miring berdimensi 3)

### Rumus Volume Paralelepiped:

$$
\text{Volume} = |\vec{a} \cdot (\vec{b} \times \vec{c})|
$$

### Penjelasan:

* $\vec{b} \times \vec{c}$ menghasilkan vektor yang tegak lurus bidang alas.
* $\vec{a} \cdot (\vec{b} \times \vec{c})$ menghitung komponen tinggi dari $\vec{a}$ terhadap bidang alas.
* Tanda mutlak (|...|) digunakan karena volume selalu positif.

### Contoh Perhitungan:

Misal:

$$
\vec{a} = \begin{bmatrix} 1 \\ 0 \\ 0 \end{bmatrix}, \quad
\vec{b} = \begin{bmatrix} 0 \\ 1 \\ 0 \end{bmatrix}, \quad
\vec{c} = \begin{bmatrix} 0 \\ 1 \\ 2 \end{bmatrix}
$$

Langkah 1: Hitung $\vec{b} \times \vec{c}$

$$
\vec{b} \times \vec{c} =
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
0 & 1 & 0 \\
0 & 1 & 2
\end{vmatrix}
= \hat{i}(1 \cdot 2 - 0 \cdot 1) - \hat{j}(0 \cdot 2 - 0 \cdot 0) + \hat{k}(0 \cdot 1 - 1 \cdot 0)
= \begin{bmatrix} 2 \\ 0 \\ 0 \end{bmatrix}
$$

Langkah 2: Hitung $\vec{a} \cdot (\vec{b} \times \vec{c})$

$$
\vec{a} \cdot \begin{bmatrix} 2 \\ 0 \\ 0 \end{bmatrix} = 1 \cdot 2 + 0 \cdot 0 + 0 \cdot 0 = 2
$$

Jadi:

$$
\text{Volume} = |\vec{a} \cdot (\vec{b} \times \vec{c})| = |2| = 2
$$

## Tugas



### 1. Luas jajaran genjang yang ditentukan oleh vektor

$$
\vec{u} = \begin{bmatrix} 1 \\ 2 \end{bmatrix}, \quad \vec{v} = \begin{bmatrix} 2 \\ 1 \end{bmatrix}
$$

Gunakan determinan untuk menghitung luas:

$$
\text{Luas} = |\vec{u} \times \vec{v}| = \left| \begin{vmatrix} 1 & 2 \\ 2 & 1 \end{vmatrix} \right| = |(1)(1) - (2)(2)| = |-3| = \boxed{3}
$$


### 2. Luas jajaran genjang yang ditentukan oleh vektor

$$
\vec{u} = \begin{bmatrix} 2 \\ 0 \end{bmatrix}, \quad \vec{v} = \begin{bmatrix} 0 \\ 3 \end{bmatrix}
$$

$$
\text{Luas} = |\vec{u} \times \vec{v}| = \left| \begin{vmatrix} 2 & 0 \\ 0 & 3 \end{vmatrix} \right| = |(2)(3) - (0)(0)| = \boxed{6}
$$


### 3. Luas segitiga dengan titik-titik sudut

$$
A = (0,0,0), \quad B = (1,3,-1), \quad C = (2,1,1)
$$

Langkah-langkah:

* Vektor AB dan AC:

   $$
   \vec{AB} = B - A = (1,3,-1), \quad \vec{AC} = C - A = (2,1,1)
   $$

* Hitung cross product $\vec{AB} \times \vec{AC}$:

$$
\vec{AB} \times \vec{AC} =
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
1 & 3 & -1 \\
2 & 1 & 1 \\
\end{vmatrix} = 
\hat{i}(3\cdot1 - (-1)\cdot1) - \hat{j}(1\cdot1 - (-1)\cdot2) + \hat{k}(1\cdot1 - 3\cdot2) = \hat{i}(4) - \hat{j}(3) + \hat{k}(-5)
$$

* Panjang vektor hasil cross product:

$$
|\vec{AB} \times \vec{AC}| = \sqrt{4^2 + (-3)^2 + (-5)^2} = \sqrt{16 + 9 + 25} = \sqrt{50}
$$

* Luas segitiga:

$$
\text{Luas} = \frac{1}{2} |\vec{AB} \times \vec{AC}| = \frac{1}{2} \sqrt{50} = \boxed{\frac{\sqrt{50}}{2}} \approx \boxed{3.54}
$$


### 4. Luas segitiga dengan titik-titik sudut

$$
A = (5,2,-1), \quad B = (3,6,2), \quad C = (1,0,4)
$$

1. Vektor AB dan AC:

   $$
   \vec{AB} = B - A = (-2, 4, 3), \quad \vec{AC} = C - A = (-4, -2, 5)
   $$

2. Cross product:

$$
\vec{AB} \times \vec{AC} =
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
-2 & 4 & 3 \\
-4 & -2 & 5 \\
\end{vmatrix}
= \hat{i}(4\cdot5 - 3\cdot(-2)) - \hat{j}(-2\cdot5 - 3\cdot(-4)) + \hat{k}(-2\cdot(-2) - 4\cdot(-4)) 
= \hat{i}(20 + 6) - \hat{j}(-10 + 12) + \hat{k}(4 + 16)
= \hat{i}(26) - \hat{j}(2) + \hat{k}(20)
$$

3. Panjang vektor hasil cross product:

$$
\sqrt{26^2 + (-2)^2 + 20^2} = \sqrt{676 + 4 + 400} = \sqrt{1080}
$$

4. Luas segitiga:

$$
\text{Luas} = \frac{1}{2} \sqrt{1080} = \boxed{\frac{\sqrt{1080}}{2}} \approx \boxed{16.43}
$$
